```python
while True:
    stuff = str(input('String to capalize [type q to quit]: '))
    if stuff=='q':
        break
    else:
        print(stuff.capitalize())
```

```python
while True:
    value=input('Integer, please [q to quit]: ')
    if value=='q':
        break
    number=ord(value)
    if number%2==0:
        continue
    else:
        print(number,'squared is',number*number)
    
```

```python
numbers=[1,3,5]
position=0
while position<len(numbers):
    number=numbers[position]
    if number%2==0:
        print('Found even number',number)
        break
    position+=1
else:#break 沒有被呼叫
    print('No even number found')
```

```python
word='thud'
offset=0
while offset<len(word):
    print(word[offset])
    offset+=1
```

```python
def document_name(func):
    def new_function(*args,**kwargs):
        print('Running function:',func.__name__)
        print('Positional arguments:',args)
        print('Keyword argument:',kwargs)
        result=func(*args,**kwargs)
        print('Result',result)
        return result
    return new_function
```

```python
def add_ints(*a):
    return sum(a) 
add_ints(3,5,4,3)
```

```python
cooler_ass_ints=document_name(add_ints)
cooler_ass_ints(3,4,5)
```

```python
@document_name
def add_ints(*a):
    return sum(a)
add_ints(3,5,4)
```

```python
def square_it(func):
    def new_function(*args,**kwargs):
        result=func(*args,**kwargs)
        return result**2
    return new_function
```

```python
@document_name
@square_it
def add_ints(a,b):
    return a+b
add_ints(3,5)
```

```python
@square_it
@document_name
def add_ints(*a):
    return sum(*a)
add_ints((1,2,3,4,5))
```

```python
def flatten(lol):
    for item in lol:
        if isinstance(item,list):
            for subitem in flatten(item):
                yield subitem
        else:
            yield item
lol=[1,2,[3,4,5],[6,[7,8,9],[]]]
flatten(lol)
```

```python
def flatten(lol):
    for item in lol:
        print('item:',item)
        if isinstance(item,list):
            print('True')
            for subitem in flatten(item):
                yield subitem
        else:
            print('False')
            yield item
lol=[1,2,[3,4,5],[6,[7,8,9],[]]]
list(flatten(lol))
```

```python
def flatten(lol):
    for item in lol:
        if isinstance(item,list):
            yield from flatten(item)
        else:
            yield item
lol=[1,2,[3,4,5],[6,[7,8,9],[]]]
tuple(flatten(lol))
```

```python
class UppercaseException(Exception):
    pass

words=['eenie','meenie','miny','MO']

try:
    for word in words:
        if word.isupper():
            raise UppercaseException(word)
except UppercaseException as e:
    print(f"偵測到大寫單字了！被抓到的單字是: {e}")
```

```python
try:
    raise UppercaseException('miny')
except UppercaseException as exc:
    print(exc)
```

```python
from collections import defaultdict
food_counter=defaultdict(int)
for food in ['spam','spam','eggs','spam']:
    food_counter[food]+=1
for food,count in food_counter.items():
    print(food,count)
```

```python
dict_counter={}
for food in ['spam','spam','eggs','spam']:
    if not food in dict_counter:
        dict_counter[food]=0
    dict_counter[food]+=1
for food,count in dict_counter.items():
    print(food,count)
```

```python
def palindrome(word):
    from collections import deque
    dq=deque(word)
    while len(dq)>1:
        if dq.popleft()!=dq.pop():
            return False
    return True
```

```python
palindrome('a')
```

```python
palindrome('racecar')
```

```python
palindrome('')
```

```python
palindrome('radar')
```

```python
palindrome('halibut')
```

```python
def another_palindrome(word):
    return word==word[::-1]
another_palindrome('radar')
```

```python
another_palindrome('halibut')
```
